# Create Event Configuration File

## Notebook Overview
This notebook builds per-event configuration files used throughout the ROMY analysis workflow.

**Main workflow**
- Read event metadata and project defaults.
- Assemble event-specific analysis parameters.
- Generate standardized config entries (time windows, channels, thresholds).
- Write configuration YAML files into event directories.

This notebook creates a configuration file for an event that is used for the analysis later.

```python
import os
import numpy as np
import matplotlib.pyplot as plt

from obspy import UTCDateTime
from sixdegrees.sixdegrees import sixdegrees
```

## Configurations

Here we load the data from a local mseed file (data_source = "mseed_file"). All parameters for the processing and plotting are set hereafter in the configuration dictionary.

In the MSEED file the station was renamed to XX.ROMY..*

```python
config = {
    # Required time window parameters
    'tbeg': "2023-12-02 14:30:00",  # Start time
    'tend': "2023-12-02 15:30:00",  # End time

    # Event information
    'origin_time': "2023-12-02 14:37:04",
    'magnitude': 7.6,

    # Station coordinates (required)
    'station_lon': 11.275476,  # BSPF longitude
    'station_lat': 48.162941,    # BSPF latitude
    
    # SEED identifiers
    'seed': "XX.ROMY..",     # Default seed ID
    'rot_seed': ["XX.ROMY..BJZ", "XX.ROMY..BJN", "XX.ROMY..BJE"], # "BW.ROMY.30.BJ*",  # Rotation sensor seed ID
    'tra_seed': ["XX.ROMY..BHZ", "XX.ROMY..BHN", "XX.ROMY..BHE"],  # Translation sensor seed ID
    
    # Data source
    'data_source': "mseed_file",  # "fdsn" | "sds" | "mseed_file"

    # Path to local mseed file
    'path_to_mseed_file': "./events/event_20231202/romy_20231202.mseed",  # "./data/eventM62_6c.mseed",

    # Path configurations
    'workdir': "./",            # Working directory
    'path_to_data_out': "./output/",  # Output directory for data
    'path_to_figures': "./events/event_20231202/figures/",  # Output directory for figures

    # Project information
    'project': "event_20231202",  # Project name
    
    # Backazimuth calculation parameters
    'baz_step': 1,               # Step size for backazimuth search in degrees
    'twin_sec': 100,          # Window length in seconds
    'twin_overlap': 0.75,  # Window overlap in seconds
    'cc_threshold': 0.8,        # Cross-correlation threshold

    # Filter parameters
    'fmin': 0.01,               # Minimum frequency for filtering
    'fmax': 0.1,                # Maximum frequency for filtering
    
    # Optional parameters
    'verbose': True,            # Print detailed information
}
```

Create a sixdegrees object

```python
sd = sixdegrees(conf=config)
```

Store the configuration as a yaml file

```python
# get date
date = config['origin_time'][:10].replace('-', '')

# create event directory
os.makedirs(f"./events/event_{date}", exist_ok=True)

# define filename
filename = f"./events/event_{date}/config_{date}.yml"

# store the configuration as a yaml file
sd.store_as_yaml(config, filename)
```
